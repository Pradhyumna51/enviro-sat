# Phase 0 — EuroSAT Dataset Sanity Check

This notebook verifies:
1. EuroSAT dataset download & extraction utility (`data.download`)
2. PyTorch dataset loading & statistics verification (`data.dataset`)
3. Class distribution visualization
4. Visual inspection of representative sample images (1 per class across all 10 land cover categories)

```python
import sys
from pathlib import Path

# Add project root to sys.path
project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import matplotlib.pyplot as plt
from data.download import download_eurosat
from data.dataset import EuroSATDataset, get_class_names, get_sample_per_class
```

## 1. Download & Verify EuroSAT Dataset

```python
raw_data_path = download_eurosat(raw_dir=str(project_root / "data" / "raw"))
print(f"Verified EuroSAT dataset at: {raw_data_path}")
```

## 2. Inspect Dataset Statistics

```python
dataset = EuroSATDataset(root_dir=raw_data_path)
stats = dataset.get_stats()

print("=" * 55)
print(f"Total Image Samples : {stats['total_samples']}")
print(f"Number of Classes   : {stats['num_classes']}")
print("=" * 55)
print("Class Distribution:")
for cls_name, count in stats['class_counts'].items():
    print(f"  - {cls_name:<22}: {count} images")
```

## 3. Class Distribution Visualization

```python
plt.figure(figsize=(12, 4.5))
classes = list(stats['class_counts'].keys())
counts = list(stats['class_counts'].values())

bars = plt.bar(classes, counts, color='#2b5c8f', edgecolor='black', alpha=0.85)
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.ylabel("Image Patch Count", fontsize=11)
plt.title("EuroSAT Land-Use Class Distribution", fontsize=14, fontweight='bold', pad=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Add values on top of bars
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 30, f'{yval}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

## 4. Sample Images Per Class (2x5 Grid)

```python
samples = get_sample_per_class(root_dir=raw_data_path)
class_names = get_class_names()

fig, axes = plt.subplots(2, 5, figsize=(15, 6.5))
fig.suptitle("EuroSAT Dataset — Sample Image Per Class (64x64 RGB)", fontsize=15, fontweight='bold')

for idx, class_name in enumerate(class_names):
    ax = axes[idx // 5, idx % 5]
    if class_name in samples:
        img = samples[class_name]
        ax.imshow(img)
        ax.set_title(f"{idx+1}. {class_name}", fontsize=11, fontweight='bold', pad=6)
    ax.axis('off')

plt.tight_layout()
plt.show()
```
